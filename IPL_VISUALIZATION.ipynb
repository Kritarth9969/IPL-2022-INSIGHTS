The Indian Premier League (IPL) is a professional T20 cricket league in India, featuring franchises representing cities. This project explores IPL 2022 match-level data to derive meaningful insights and understand match outcomes, player performances, and team dynamics.


```python
# Load the libraries for meaning full insights
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# for removing some warning we use the warning like...
import warnings
warnings.filterwarnings('ignore')
```

```python
# Which team  won the most matches..?
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index,x = match_wins.values,palette='viridis')
plt.title('Most Match win by team')
plt.xlabel('Number of Matches')
plt.ylabel('Team Name')
plt.savefig('Most Match win by team.png',dpi=300)
plt.show()
```

```python
# Toss decision Trend
# for fielding and batting
sns.countplot(x = df['toss_decision'], palette = 'Set2')
plt.title('Toss Decision Trend')
plt.xlabel('Toss Decision')
plt.ylabel('Count')
plt.savefig('Toss Decision Trend.png',dpi=300)
plt.show()
```

```python
# toss winner vs match winner data 
df[df['toss_winner'] == df['match_winner']]
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count/df.shape[0])*100
percentage
```

```python
# How do team win (run vs wickets)
sns.countplot(x = df['won_by'],palette = 'Set1')
plt.title("Won By Wicket or Runs")
plt.xlabel('Won By')
plt.ylabel('Count')
plt.savefig('Won By Wicket or Runs.png',dpi=300)
plt.show()
```

```python
# Most player of the match award

count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x = count.values,y = count.index,palette='Set2')
plt.title('Most Player of the Match Award')
plt.xlabel('Number of Awards')
plt.ylabel('Player Name')
plt.savefig('Most Player of the Match Award.png',dpi=300)
plt.show()
```

```python
# Top 3 scorers
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(3)
high
```

```python
high.plot(kind = 'barh',color = 'orange')
plt.title('Top 3 Scorers')
plt.xlabel('Total Runs')
plt.ylabel('Player Name')
plt.savefig('Top 3 Scorers.png',dpi=300)
plt.show()
```

```python
# 10 best bowling figures
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
# Combined visualization cell for existing variables: df, match_wins, count, high, top_bowlers, venue
sns.set_style('whitegrid')
fig, axes = plt.subplots(4, 2, figsize=(14, 18))
axes = axes.flatten()

# 1. Most matches won by team
sns.barplot(x=match_wins.values, y=match_wins.index, palette='viridis', ax=axes[0])
axes[0].set_title('Most Matches Won')
axes[0].set_xlabel('Number of Matches')
axes[0].set_ylabel('Team')

# 2. Toss decision trend
sns.countplot(x=df['toss_decision'], palette='Set2', ax=axes[1])
axes[1].set_title('Toss Decision Trend')
axes[1].set_xlabel('')
axes[1].set_ylabel('Count')

# 3. Toss winner vs match winner (pie)
tf_bool = (df['toss_winner'] == df['match_winner'])
tf_counts = pd.Series({'Yes': int(tf_bool.sum()), 'No': int((~tf_bool).sum())})
axes[2].pie(tf_counts.values, labels=tf_counts.index, autopct='%1.1f%%', colors=['#66b3ff', '#ff9999'])
axes[2].set_title('Did Toss Winner Also Win Match?')
axes[2].axis('equal')

# 4. Won by (Runs vs Wickets)
sns.countplot(x=df['won_by'], palette='Set1', ax=axes[3])
axes[3].set_title('Matches Won by Runs vs Wickets')
axes[3].set_xlabel('')
axes[3].set_ylabel('Count')

# 5. Top 10 Player of the Match awards
sns.barplot(x=count.values, y=count.index, palette='Set3', ax=axes[4])
axes[4].set_title('Top 10 Player of the Match Awards')
axes[4].set_xlabel('Number of Awards')
axes[4].set_ylabel('Player')

# 6. Top 3 scorers (horizontal bar)
high.plot(kind='barh', color='orange', ax=axes[5])
axes[5].set_title('Top 3 Scorers (Total Runs)')
axes[5].set_xlabel('Total Runs')
axes[5].set_ylabel('Player')

# 7. Top 10 bowlers by wickets (horizontal bar)
top_bowlers.plot(kind='barh', color='teal', ax=axes[6])
axes[6].set_title('Top Bowlers (Wickets)')
axes[6].set_xlabel('Wickets')
axes[6].set_ylabel('Player')

# 8. Most matches by venue (top)
sns.barplot(x=venue.values, y=venue.index, palette='magma', ax=axes[7])
axes[7].set_title('Most Matches Played by Venue')
axes[7].set_xlabel('Number of Matches')
axes[7].set_ylabel('Venue')

plt.tight_layout()
plt.savefig('IPL_Combined_Visualization.png', dpi=300)
plt.show()
```
